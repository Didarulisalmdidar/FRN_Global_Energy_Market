```python
import pandas as pd
import numpy as np

for i in range(1, 13):
    # Load the CSV file
    df = pd.read_csv('window_{}.csv'.format(i))
    dataset_return = pd.DataFrame()
    
    # Ensure columns are numeric and compute log returns
    for col_name in df.columns:
        try:
            df[col_name] = pd.to_numeric(df[col_name], errors='coerce')  # Convert to numeric
            dataset_return[col_name] = np.log(df[col_name]) - np.log(df[col_name].shift(1))
        except Exception as e:
            print(f"Error processing column {col_name} in file window_{i}.csv: {e}")
    
    # Drop rows with NaN values
    dataset_return = dataset_return.dropna()
    
    # Save the processed DataFrame
    dataset_return.to_csv('log_window_{}.csv'.format(i))
```
