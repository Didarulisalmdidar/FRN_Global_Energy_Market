```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor

for i in range(1, 13):
    df = pd.read_csv(f'log_window_{i}.csv', index_col=0)
    
    feature_importance_scores = {}
    
    for feature_name in df.columns:
        y = df[feature_name].iloc[1:]  # dependent variable
        x = df.drop(df.index[-1])     # independent variables
        
        model = RandomForestRegressor(n_estimators=200, max_features='log2', random_state=42)
        model.fit(x, y)
        
        score = model.feature_importances_
        
        # Invert the importance scores using (max - score)
        inverted = score.max() - score
        
        # Normalize to make them sum to 1
        if inverted.sum() > 0:
            inverted_normalized = inverted / inverted.sum()
        else:
            inverted_normalized = np.zeros_like(inverted)  # fallback in case of all zeros
        
        feature_importance_scores[feature_name] = inverted_normalized
    
    # Convert dict to matrix
    importance_matrix = [v for v in feature_importance_scores.values()]
    importance_matrix = np.array(importance_matrix)
    
    # Save to DataFrame
    feature_names = list(feature_importance_scores.keys())
    feature_matrix = pd.DataFrame(importance_matrix, index=feature_names, columns=df.columns)
    feature_matrix.to_csv(f'FMin{i}.csv')
```
