```python
import os
import numpy as np
import pandas as pd
df=pd.read_csv('TMin2.csv',index_col=0)
oil_gas=df.columns[[0, 2, 7, 8, 10, 14, 17, 21, 23, 29, 30, 32, 34,38]]
Electric=df.columns[[1,3, 5, 11, 15, 19, 24, 26]]
Infrastructure=df.columns[[12, 16, 18, 25, 35, 37]]
Renewable=df.columns[[4, 6, 9, 13, 20, 22, 27, 28, 33, 36,31]]
```

```python
import numpy as np
import pandas as pd

sectors = {
    "Oil_Gas": oil_gas,
    "Electric": Electric,
    "Infrastructure": Infrastructure,
    "Renewable": Renewable,
}
names = list(sectors.keys())

def build_influence_matrix(df, normalize=True):
    n = len(names)
    IN = np.zeros((n, n))
    for a, m in enumerate(names):          # m = influencing sector
        for b, nsec in enumerate(names):   # n = influenced sector
            if m == nsec:
                continue
            total = 0
            for col in sectors[m]:
                total += df[col][sectors[nsec]].sum()
            if normalize:
                total = total / (len(sectors[m]) * len(sectors[nsec]))
            IN[a, b] = total
    return IN

df = pd.read_csv('TMin11.csv', index_col=0)
In = build_influence_matrix(df, normalize=True)

print(pd.DataFrame(In, index=names, columns=names).round(4))
```

```python
def build_di_matrix(In):
    n = len(In)
    DI = [[0]*n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            if i != j:
                diff = In[i][j] - In[j][i]
                DI[i][j] = diff if diff > 0 else 0
            else:
                DI[i][j] = 0  # No self influence
    return DI

di_matrix = build_di_matrix(In)

# Display DI matrix
for row in di_matrix:
    print(row)

# Labeled version (easier to read)
print(pd.DataFrame(di_matrix, index=names, columns=names).round(4))
```

```python
import networkx as nx
import matplotlib.pyplot as plt

labels = ['Oil_GAS', 'Electric', 'Infrastructure', 'Renewable']

G = nx.DiGraph()
G.add_nodes_from(labels)

all_weights = [di_matrix[i][j] for i in range(4) for j in range(4) if i != j]
max_weight = max(all_weights) if all_weights else 1

for i in range(4):
    for j in range(4):
        if i != j:
            influence = di_matrix[i][j]
            norm_width = 0.5 if influence == 0 else 1 + (influence / max_weight) * 5
            G.add_edge(labels[i], labels[j], weight=influence, width=norm_width)

pos = nx.circular_layout(G)

nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=2000)
nx.draw_networkx_labels(G, pos, font_size=6, font_weight='bold')

for u, v, data in G.edges(data=True):
    weight = data['weight']
    width = data['width']
    if weight > 0:
        edge_color = 'black'
        style = 'solid'
        arrowsize = 30
        edge_width = max(width, 2)
    else:
        edge_color = 'gray'
        style = 'dotted'
        arrowsize = 30
        edge_width = 1

    nx.draw_networkx_edges(
        G, pos,
        edgelist=[(u, v)],
        width=edge_width,
        arrowstyle='-|>',
        arrowsize=arrowsize,
        edge_color=edge_color,
        style=style,
        connectionstyle='arc3,rad=0.3',
        min_source_margin=15,
        min_target_margin=15
    )

plt.title("(d) May-Aug 2023", fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.savefig('may-aug23Manu.png')
plt.show()
```
