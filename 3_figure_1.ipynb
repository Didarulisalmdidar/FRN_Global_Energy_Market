```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
for i in range(1, 13):
    # Read the CSV file
    df = pd.read_csv(f'FMin{i}.csv', index_col=0)
    df.columns = company_list  # Rename columns to match company_list
    
    # Sum each column and add to the total_sums dictionar
    
    for company in company_list:
        company_sums[company].append(df[company].sum())
        

for company, sums in company_sums.items():
    print(f"Sums for {company}: {sums}")   
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.interpolate import make_interp_spline

sns.set(style="whitegrid")

selected_indices = [0, 2, 7, 14, 21, 30, 34, 38]
selected_companies = [company_list[i] for i in selected_indices]

x_labels = [
    "Jan-Apr20", "May-Aug20", "Sep-Dec20", "Jan-Apr21", "May-Aug21", "Sep-Dec21",
    "Jan-Apr22", "May-Aug22", "Sep-Dec22", "Jan-Apr23", "May-Aug23", "Sep-Dec23"
]

plt.figure(figsize=(14, 7), dpi=120)

colors = sns.color_palette("tab10")
line_styles = ['-', '--', '-.', ':', '-']
markers = ['o', 's', 'D', '^', 'v', '<', '>', 'P']  # Different marker styles

for i, company in enumerate(selected_companies):
    x = np.arange(len(company_sums[company]))
    y = company_sums[company]

    spline = make_interp_spline(x, y, k=3)
    x_new = np.linspace(x.min(), x.max(), 300)
    y_new = spline(x_new)

    # Plot smooth lines
    plt.plot(
        x_new, y_new,
        label=company,
        linestyle=line_styles[i % len(line_styles)],
        color=colors[i % len(colors)],
        linewidth=3,
        alpha=0.9
    )

    # Plot markers on original points with different marker shapes
    plt.scatter(
        x, y,
        marker=markers[i % len(markers)],
        s=120,
        edgecolor=colors[i % len(colors)],
        facecolor='white',
        linewidth=2.5,
        zorder=5,
        alpha=1.0
    )

plt.xlabel('Time Segment', fontsize=16, fontweight='bold', family='Arial')
plt.ylabel('Influence', fontsize=16, fontweight='bold', family='Arial')
plt.title('(a) Influence of Oil and Gas Energy Companies Over Time', fontsize=18, fontweight='bold', family='Arial')

plt.xticks(ticks=range(12), labels=x_labels, rotation=45, fontsize=12, fontweight='bold', family='Arial')
plt.yticks(fontsize=12, fontweight='bold', family='Arial')

plt.grid(True, linestyle='--', linewidth=0.6, alpha=0.5)

plt.legend(
    title='Companies',
    fontsize=12,
    title_fontsize=14,
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    frameon=False
)

plt.tight_layout()
plt.gcf().set_facecolor('#f7f7f7')

plt.savefig('Oil_and_Gas_FinalRev.png', bbox_inches='tight')
plt.show()
```
