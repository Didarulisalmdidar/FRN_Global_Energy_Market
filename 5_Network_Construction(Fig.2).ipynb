```python
import networkx as nx
G=nx.DiGraph()
G.add_nodes_from(df.index)
cnt=0
for source in df.index:
    for target in df.columns:
        if df.loc[source, target] == 1:
            G.add_edge(target,source)
import matplotlib.pyplot as plt
import networkx as nx

# Position layout
pos = nx.spring_layout(G, k=2.95)
plt.figure(figsize=(25, 20))

# Prepare node colors: half lightgreen, half lightblue
nodes = list(G.nodes())
half = len(nodes) // 2
node_colors = ['cyan'] * half + ['lightblue'] * (len(nodes) - half)

# Draw the graph with customized node colors
nx.draw(
    G, pos,
    with_labels=True,
    node_size=8000,
    node_color=node_colors,
    edge_color='black',
    arrows=True,
    arrowsize=15,
    font_color='red',
    font_size=16,
    font_weight='bold'
)

plt.tight_layout()
plt.savefig('network_in.png')
plt.show()
```
